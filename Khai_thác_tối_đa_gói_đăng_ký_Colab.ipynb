## GPU nhanh hơn

<p>Người dùng đã mua một trong các gói trả phí của Colab có quyền sử dụng GPU có tốc độ nhanh hơn và bộ nhớ có dung lượng lớn hơn. Bạn có thể nâng cấp chế độ cài đặt GPU của sổ tay bằng cách chuyển đến mục <code>Thời gian chạy &gt; Thay đổi loại thời gian chạy</code> trong trình đơn để chọn trong số nhiều lựa chọn trình tăng tốc, tuỳ thuộc vào phạm vi cung cấp.</p>
<p>Khi chọn phiên bản Colab miễn phí, bạn sẽ có quyền dùng GPU T4 của Nvidia theo phạm vi cung cấp và các giới hạn về hạn mức.</p>

Bạn có thể xem GPU mình đã chỉ định bất cứ lúc nào bằng cách thực thi ô sau. Nếu kết quả thực thi là "Not connected to a GPU" sau khi chạy ô chứa mã ở bên dưới, thì bạn có thể thay đổi môi trường thời gian chạy bằng cách chuyển đến mục <code>Thời gian chạy &gt; Thay đổi loại thời gian chạy</code> trong trình đơn để bật trình tăng tốc GPU, sau đó thực thi lại ô chứa mã.

```
gpu_info = !nvidia-smi
gpu_info = '\n'.join(gpu_info)
if gpu_info.find('failed') >= 0:
  print('Not connected to a GPU')
else:
  print(gpu_info)
```

Để dùng một GPU với sổ tay của bạn, hãy chọn trình đơn <code>Thời gian chạy &gt; Thay đổi loại thời gian chạy</code> rồi chọn trình tăng tốc phần cứng mong muốn.

## Bộ nhớ lớn hơn

Người dùng đã mua một trong các gói trả phí của Colab được quyền sử dụng máy ảo có dung lượng bộ nhớ lớn khi có sẵn. GPU mạnh hơn luôn đi kèm với máy ảo có dung lượng bộ nhớ lớn.
Bạn có thể xem dung lượng bộ nhớ hiện tại của mình bất cứ lúc nào bằng cách chạy ô chứa mã sau. Nếu kết quả thực thi là "Not using a high-RAM runtime" sau khi chạy ô chứa mã ở bên dưới, thì bạn có thể bật môi trường thời gian chạy có RAM cao cấp qua mục <code>Thời gian chạy &gt; Thay đổi loại thời gian chạy</code> trong trình đơn. Tiếp đến, hãy chọn RAM cao cấp trong nút bật/tắt Cấu hình môi trường thời gian chạy. Sau đó, thực thi lại ô chứa mã.

```
import psutil

ram_gb = psutil.virtual_memory().total / 1e9
print('Your runtime has {:.1f} gigabytes of available RAM\n'.format(ram_gb))

if ram_gb < 20:
  print('Not using a high-RAM runtime')
else:
  print('You are using a high-RAM runtime!')
```
